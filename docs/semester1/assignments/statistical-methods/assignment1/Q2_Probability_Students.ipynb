# Q2: Probability Problem - Students, Sports, and Grades

**Problem Statement:**
In a class of 40 students, there are 18 boys and 22 girls. 
- Out of the boys: 7 participate in sports, 6 scored an 'A' grade, with 3 boys involved in both.
- Among the girls: 9 participate in sports, 8 scored an 'A' grade, and 4 girls did both.

**Question:** If a student is picked at random, what is the probability the student is either involved in sports or scored an 'A' grade?


```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.patches import Rectangle

# Set style
try:
    plt.style.use('seaborn-v0_8')
except:
    try:
        plt.style.use('seaborn')
    except:
        plt.style.use('default')
```

```python
# Given data
total_students = 40
boys = 18
girls = 22

# Boys data
boys_sports = 7
boys_grade_A = 6
boys_both = 3

# Girls data
girls_sports = 9
girls_grade_A = 8
girls_both = 4

print("=" * 60)
print("GIVEN DATA")
print("=" * 60)
print(f"Total students: {total_students}")
print(f"\nBoys: {boys}")
print(f"  - Boys in sports: {boys_sports}")
print(f"  - Boys with 'A' grade: {boys_grade_A}")
print(f"  - Boys in both: {boys_both}")
print(f"\nGirls: {girls}")
print(f"  - Girls in sports: {girls_sports}")
print(f"  - Girls with 'A' grade: {girls_grade_A}")
print(f"  - Girls in both: {girls_both}")
print("=" * 60)
```

## Calculate Individual Sets


```python
# Calculate boys only in sports (not in both)
boys_sports_only = boys_sports - boys_both

# Calculate boys only with A grade (not in both)
boys_grade_A_only = boys_grade_A - boys_both

# Calculate boys in neither
boys_neither = boys - boys_sports - boys_grade_A + boys_both

# Calculate girls only in sports (not in both)
girls_sports_only = girls_sports - girls_both

# Calculate girls only with A grade (not in both)
girls_grade_A_only = girls_grade_A - girls_both

# Calculate girls in neither
girls_neither = girls - girls_sports - girls_grade_A + girls_both

# Total students in sports
total_sports = boys_sports + girls_sports

# Total students with A grade
total_grade_A = boys_grade_A + girls_grade_A

# Total students in both
total_both = boys_both + girls_both

# Total students in sports OR grade A (using inclusion-exclusion principle)
total_sports_or_grade_A = total_sports + total_grade_A - total_both

print("=" * 60)
print("BREAKDOWN BY CATEGORY")
print("=" * 60)
print("\nBOYS BREAKDOWN:")
print(f"  - Sports only: {boys_sports_only}")
print(f"  - Grade A only: {boys_grade_A_only}")
print(f"  - Both: {boys_both}")
print(f"  - Neither: {boys_neither}")
print(f"  - Total boys: {boys_sports_only + boys_grade_A_only + boys_both + boys_neither}")

print("\nGIRLS BREAKDOWN:")
print(f"  - Sports only: {girls_sports_only}")
print(f"  - Grade A only: {girls_grade_A_only}")
print(f"  - Both: {girls_both}")
print(f"  - Neither: {girls_neither}")
print(f"  - Total girls: {girls_sports_only + girls_grade_A_only + girls_both + girls_neither}")

print("\n" + "=" * 60)
print("TOTAL COUNTS")
print("=" * 60)
print(f"Total students in sports: {total_sports}")
print(f"Total students with 'A' grade: {total_grade_A}")
print(f"Total students in both: {total_both}")
print(f"\nTotal students in sports OR 'A' grade: {total_sports_or_grade_A}")
print("=" * 60)
```

## Solution: Using Inclusion-Exclusion Principle

**Formula:** P(Sports OR Grade A) = P(Sports) + P(Grade A) - P(Sports AND Grade A)


```python
# Calculate probabilities
P_sports = total_sports / total_students
P_grade_A = total_grade_A / total_students
P_both = total_both / total_students

# Using inclusion-exclusion principle: P(A ∪ B) = P(A) + P(B) - P(A ∩ B)
P_sports_or_grade_A = P_sports + P_grade_A - P_both

# Alternative calculation (direct count)
P_sports_or_grade_A_direct = total_sports_or_grade_A / total_students

print("=" * 60)
print("PROBABILITY CALCULATION")
print("=" * 60)
print(f"P(Sports) = {total_sports}/{total_students} = {P_sports:.4f}")
print(f"P(Grade A) = {total_grade_A}/{total_students} = {P_grade_A:.4f}")
print(f"P(Sports AND Grade A) = {total_both}/{total_students} = {P_both:.4f}")
print("\n" + "-" * 60)
print("Using Inclusion-Exclusion Principle:")
print(f"P(Sports OR Grade A) = P(Sports) + P(Grade A) - P(Sports AND Grade A)")
print(f"P(Sports OR Grade A) = {P_sports:.4f} + {P_grade_A:.4f} - {P_both:.4f}")
print(f"P(Sports OR Grade A) = {P_sports_or_grade_A:.4f}")
print("\n" + "-" * 60)
print("Direct Calculation:")
print(f"P(Sports OR Grade A) = {total_sports_or_grade_A}/{total_students} = {P_sports_or_grade_A_direct:.4f}")
print("=" * 60)
print(f"\n✅ FINAL ANSWER: {P_sports_or_grade_A:.4f} or {P_sports_or_grade_A*100:.2f}%")
print("=" * 60)
```

## Visualization


```python
# Create visualization
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Venn Diagram representation (simplified bar chart)
categories = ['Sports\nOnly', 'Grade A\nOnly', 'Both', 'Neither']
boys_counts = [boys_sports_only, boys_grade_A_only, boys_both, boys_neither]
girls_counts = [girls_sports_only, girls_grade_A_only, girls_both, girls_neither]

x = np.arange(len(categories))
width = 0.35

axes[0].bar(x - width/2, boys_counts, width, label='Boys', color='lightblue', edgecolor='black')
axes[0].bar(x + width/2, girls_counts, width, label='Girls', color='lightpink', edgecolor='black')
axes[0].set_xlabel('Category', fontsize=12)
axes[0].set_ylabel('Number of Students', fontsize=12)
axes[0].set_title('Distribution of Students by Category', fontsize=14, fontweight='bold')
axes[0].set_xticks(x)
axes[0].set_xticklabels(categories)
axes[0].legend()
axes[0].grid(True, alpha=0.3, axis='y')

# Add value labels on bars
for i, (b, g) in enumerate(zip(boys_counts, girls_counts)):
    axes[0].text(i - width/2, b + 0.2, str(b), ha='center', va='bottom', fontweight='bold')
    axes[0].text(i + width/2, g + 0.2, str(g), ha='center', va='bottom', fontweight='bold')

# Probability breakdown
prob_categories = ['P(Sports)', 'P(Grade A)', 'P(Both)', 'P(Sports OR Grade A)']
prob_values = [P_sports, P_grade_A, P_both, P_sports_or_grade_A]
colors = ['skyblue', 'lightgreen', 'orange', 'red']

bars = axes[1].bar(prob_categories, prob_values, color=colors, edgecolor='black', alpha=0.7)
axes[1].set_ylabel('Probability', fontsize=12)
axes[1].set_title('Probability Breakdown', fontsize=14, fontweight='bold')
axes[1].set_ylim(0, max(prob_values) * 1.2)
axes[1].grid(True, alpha=0.3, axis='y')

# Add value labels on bars
for bar, val in zip(bars, prob_values):
    height = bar.get_height()
    axes[1].text(bar.get_x() + bar.get_width()/2., height + 0.01,
                f'{val:.3f}', ha='center', va='bottom', fontweight='bold', fontsize=11)

plt.tight_layout()
plt.show()

# Summary table
summary_data = {
    'Category': ['Total Students', 'In Sports', "With 'A' Grade", 'In Both', 'In Sports OR Grade A'],
    'Count': [total_students, total_sports, total_grade_A, total_both, total_sports_or_grade_A],
    'Probability': [1.0, P_sports, P_grade_A, P_both, P_sports_or_grade_A]
}
summary_df = pd.DataFrame(summary_data)
summary_df['Percentage'] = summary_df['Probability'] * 100

print("\nSUMMARY TABLE:")
print(summary_df.to_string(index=False))
```
